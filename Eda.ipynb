```python
numeric_cols = [
    'company_size_employees',
    'years_experience_min',
    'years_experience_max',
    'number_of_applicants',
    'number_of_interview_rounds',
    'offers_extended',
    'offers_accepted',
    'salary_min_usd',
    'salary_max_usd'
]

df[numeric_cols].describe().T
```

```python
def clean_salary(x):
    return pd.to_numeric(
        x.astype(str).str.replace(r'[^0-9.]', '', regex=True),
        errors='coerce'
    )

df['salary_min_usd'] = clean_salary(df['salary_min_usd'])
df['salary_max_usd'] = clean_salary(df['salary_max_usd'])

# Convert invalid negative salaries to missing
df.loc[df['salary_min_usd'] < 0, 'salary_min_usd'] = np.nan
df.loc[df['salary_max_usd'] < 0, 'salary_max_usd'] = np.nan
```

```python
df['avg_salary'] = (
    df['salary_min_usd'] + df['salary_max_usd']
) / 2
df['avg_salary'].describe()
```

```python

plt.figure(figsize=(10, 5))

sns.histplot(
    df['avg_salary'].dropna(),
    bins=40,

)

plt.title('Distribution of Average Job Salary')
plt.xlabel('Average Salary (USD)')
plt.ylabel('Number of Jobs')
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    df['number_of_applicants'].dropna(),
    bins=40
)

plt.title('Distribution of Number of Applicants')
plt.xlabel('Number of Applicants')
plt.ylabel('Number of Jobs')
plt.show()
```
