```python
numeric_cols = [
    'company_size_employees',
    'years_experience_min',
    'years_experience_max',
    'number_of_applicants',
    'number_of_interview_rounds',
    'offers_extended',
    'offers_accepted',
    'salary_min_usd',
    'salary_max_usd'
]

df[numeric_cols].describe().T
```

```python
def clean_salary(x):
    return pd.to_numeric(
        x.astype(str).str.replace(r'[^0-9.]', '', regex=True),
        errors='coerce'
    )

df['salary_min_usd'] = clean_salary(df['salary_min_usd'])
df['salary_max_usd'] = clean_salary(df['salary_max_usd'])

# Convert invalid negative salaries to missing
df.loc[df['salary_min_usd'] < 0, 'salary_min_usd'] = np.nan
df.loc[df['salary_max_usd'] < 0, 'salary_max_usd'] = np.nan
df.loc[df["salary_max_usd"] < df["salary_min_usd"], ("salary_max_usd","salary_min_usd")] = np.nan
```

# Feature Engineering

```python
df['avg_salary'] = (
    df['salary_min_usd'] + df['salary_max_usd']
) / 2
df['avg_salary'].describe()
```

## Outliers detection

```python
numeric_cols = df.select_dtypes(include='number').columns
outlier_summary = {}

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]
    
    outlier_summary[col] = {
        
        'Outlier Count': len(outliers),
        'Outlier Percentage': round((len(outliers) / df[col].notna().sum()) * 100, 2)
    }

outlier_summary
```

```python
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    plt.figure(figsize=(8, 4))
    plt.boxplot(df[col].dropna())
    plt.title(f'Boxplot of {col}')
    plt.xlabel(col)
    plt.show()
```

Since each numerical col has a lot of outliers and most of them are valid so we will not remove them
